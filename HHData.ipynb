*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
def city_feature(x: str) -> tuple:
    million_cities = ['Новосибирск', 'Екатеринбург','Нижний Новгород','Казань', 'Челябинск','Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 'Воронеж','Волгоград']
    list_x = x.split(' , ') # Надеюсь, что в данных гарантировано, что разделение
                            # информации, которая не в скобках именно запятая с пробелами по бокам.
    city = list_x[0]
    if city in ['Москва', 'Санкт-Петербург']:
        city = city
    elif city in million_cities:
        city = 'город-миллионик'
    else: city = 'другие'
    list_x.pop(0)
    move = -1
    businessTrip = -1
    for i in range(len(list_x)):
        if 'пере' in list_x[i]:
            move = i
        if 'команд' in list_x[i]:
            businessTrip = i
    if move != -1:
        if list_x[move].split(' ')[0].lower() == 'не':
            readyToMove = False
        else:
            readyToMove = True
    else: readyToMove = False
    if businessTrip != -1:
        if list_x[businessTrip].split(' ')[0].lower() == 'не':
            readyToBusinessTrip = False
        else:
            readyToBusinessTrip = True
    else: readyToBusinessTrip = False
    return city, readyToMove, readyToBusinessTrip

hh_data['Город, переезд, командировки'].isna().sum() # Проверка на NaN в столбце, т.к у нас их нет, в функции они не учитываются. 

hh_data['1']= hh_data['Город, переезд, командировки'].apply(lambda x: city_feature(x))

hh_data['Город'] = hh_data['1'].apply(lambda x: x[0])
hh_data['Готовность к переезду'] = hh_data['1'].apply(lambda x: x[1])
hh_data['Готовность к командировкам'] = hh_data['1'].apply(lambda x: x[2]) # Долго думал как это всё распаковать, в итоге решил так, 
                                                                            # Не нужно вызывать много раз функцию, где циклы и т.д, а просто распределение значений.
hh_data = hh_data.drop(columns=['1', 'Город, переезд, командировки'])      
print(f'Процент жителей в Питере: {((hh_data['Город'].value_counts(normalize=True)['Санкт-Петербург'])*100).round()}%')
print('Процент, кто готов к командировкам и переездам одновременно: ', ((((hh_data['Готовность к командировкам']) & (hh_data['Готовность к переезду'])).sum() / len(hh_data)) * 100).round())


```

4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".

```python
exchange = pd.read_csv('Data/ExchangeRates.csv')

hh_data['Обновление резюме'] = pd.to_datetime(hh_data['Обновление резюме'])
exchange['date'] = pd.to_datetime(exchange['date'])
currency_ISO = {'грн': 'UAH', 'USD':'USD', 'EUR':'EUR', 'белруб':'BYN', 'KGS':'KGS', 'сум':'UZS', 'AZN':'AZN', 'KZT':'KZT', 'руб':'RUB'}
hh_data['salary'] = hh_data['ЗП'].apply(lambda x: x.split(' ')[0])
hh_data['currency'] = hh_data['ЗП'].apply(lambda x: currency_ISO[x.split(' ')[1].strip().replace('.','')])


exchange = exchange.drop(columns=['per','time', 'vol'])
exchange['new_date'] = exchange['date'].dt.date

hh_data['new_date'] = hh_data['Обновление резюме'].dt.date
hh_data = hh_data.merge(exchange, left_on=['new_date','currency'], right_on=['new_date', 'currency'], how='left')

hh_data['close'] = hh_data['close'].fillna(1)
hh_data['proportion'] = hh_data['proportion'].fillna(1)

hh_data = hh_data.astype({'salary':'float', 'close': 'float', 'proportion':'float'})

hh_data['ЗП(в руб)'] = ((hh_data.close / hh_data.proportion) * hh_data.salary).astype('float')


```

В какую сторону асимметрично логарифмическое распределение? Напишите об этом в комментарии к графику.
Найдите выбросы с помощью метода z-отклонения и удалите их из данных, используйте логарифмический масштаб. Давайте сделаем послабление на **1 сигму** (возьмите 4 сигмы) в **правую сторону**.

Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?

```python
fig = plt.figure(figsize=(8,6))
log_age = np.log(hh_data['Возраст'])
histplot = sns.histplot(log_age, bins=50)
mean = log_age.mean()
std = log_age.std()
histplot.axvline(mean, color='k', lw=2)

histplot.axvline(mean - 3 * std, color='k', lw=2, linestyle='--')

histplot.axvline(mean + 3 * std, color='k', lw=2, linestyle='--')

histplot.set_title('Логарифмическое распределение возраста')
histplot.set_ylabel('Количество человек')

def z_score_outliers(data, feature, log_scale=False, left=3, right=3):
    if log_scale:
        x = np.log(data[feature])
    else:
        x = data[feature]
    mean = x.mean()
    std = x.std()
    
    lower_bound = mean - (left * std)
    upper_bound = mean + (right * std)
    
    cleaned = data[(x > lower_bound) & (x < upper_bound)]
    outliers = data[(x < lower_bound) | (x > upper_bound)]
    return outliers, cleaned

outliers, cleaned = z_score_outliers(hh_data, 'Возраст', log_scale=True, right=4)

hh_data = cleaned

outliers.head()


```

На графике видно правостороннюю ассиметрию, у график идёт больший хвост в правую сторону. Это вполне логично, соискателей младшего возраста быть не может, меньше 16 лет не должно быть наблюдений. А в правую сторону ограничений почти нет, люди могут искать работу и в 60-80 лет.
